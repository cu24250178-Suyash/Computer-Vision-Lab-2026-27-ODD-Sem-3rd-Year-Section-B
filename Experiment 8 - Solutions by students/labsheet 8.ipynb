```python
import cv2
import numpy as np

cap = cv2.VideoCapture("video.mp4")

ret, frame = cap.read()

gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

corners = cv2.goodFeaturesToTrack(
    gray,
    maxCorners=100,
    qualityLevel=0.3,
    minDistance=7
)

if corners is not None:
    corners = np.intp(corners)

    for corner in corners:
        x, y = corner.ravel()
        cv2.circle(frame, (x, y), 5, (0, 255, 0), -1)

cv2.imshow("Shi-Tomasi Corners", frame)
cv2.waitKey(0)

cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

cap = cv2.VideoCapture("video.mp4")

ret, old_frame = cap.read()
old_gray = cv2.cvtColor(old_frame, cv2.COLOR_BGR2GRAY)

p0 = cv2.goodFeaturesToTrack(
    old_gray,
    maxCorners=100,
    qualityLevel=0.3,
    minDistance=7
)

while True:

    ret, frame = cap.read()

    if not ret:
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    p1, status, error = cv2.calcOpticalFlowPyrLK(
        old_gray,
        gray,
        p0,
        None,
        winSize=(15, 15),
        maxLevel=2,
        criteria=(
            cv2.TERM_CRITERIA_EPS |
            cv2.TERM_CRITERIA_COUNT,
            10,
            0.03
        )
    )

    if p1 is not None:

        good_new = p1[status == 1]
        good_old = p0[status == 1]

        for new, old in zip(good_new, good_old):

            x_new, y_new = new.ravel()
            x_old, y_old = old.ravel()

            cv2.arrowedLine(
                frame,
                (int(x_old), int(y_old)),
                (int(x_new), int(y_new)),
                (0, 255, 0),
                2
            )

            cv2.circle(
                frame,
                (int(x_new), int(y_new)),
                4,
                (0, 0, 255),
                -1
            )

        p0 = good_new.reshape(-1, 1, 2)

    cv2.imshow("Lucas-Kanade Tracking", frame)

    old_gray = gray.copy()

    if cv2.waitKey(30) & 0xFF == 27:
        break

cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

cap = cv2.VideoCapture("video.mp4")

ret, old_frame = cap.read()
old_gray = cv2.cvtColor(old_frame, cv2.COLOR_BGR2GRAY)

while True:

    ret, frame = cap.read()

    if not ret:
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    flow = cv2.calcOpticalFlowFarneback(
        old_gray,
        gray,
        None,
        0.5,
        3,
        15,
        3,
        5,
        1.2,
        0
    )

    magnitude, angle = cv2.cartToPolar(
        flow[..., 0],
        flow[..., 1]
    )

    hsv = np.zeros_like(frame)

    hsv[..., 1] = 255

    hsv[..., 0] = angle * 180 / np.pi / 2

    hsv[..., 2] = cv2.normalize(
        magnitude,
        None,
        0,
        255,
        cv2.NORM_MINMAX
    )

    result = cv2.cvtColor(
        hsv,
        cv2.COLOR_HSV2BGR
    )

    cv2.imshow("Farneback Dense Optical Flow", result)

    old_gray = gray.copy()

    if cv2.waitKey(30) & 0xFF == 27:
        break

cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

cap = cv2.VideoCapture("video.mp4")

ret, old_frame = cap.read()
old_gray = cv2.cvtColor(old_frame, cv2.COLOR_BGR2GRAY)

p0 = cv2.goodFeaturesToTrack(
    old_gray,
    maxCorners=50,
    qualityLevel=0.3,
    minDistance=7
)

total_distance = 0
frame_count = 0

while True:

    ret, frame = cap.read()

    if not ret:
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    p1, status, error = cv2.calcOpticalFlowPyrLK(
        old_gray,
        gray,
        p0,
        None
    )

    if p1 is not None:

        good_new = p1[status == 1]
        good_old = p0[status == 1]

        for new, old in zip(good_new, good_old):

            x_new, y_new = new.ravel()
            x_old, y_old = old.ravel()

            distance = np.sqrt(
                (x_new - x_old)**2 +
                (y_new - y_old)**2
            )

            total_distance += distance

        frame_count += 1
        p0 = good_new.reshape(-1, 1, 2)

    old_gray = gray.copy()

    if cv2.waitKey(1) & 0xFF == 27:
        break

cap.release()

if frame_count > 0:
    average_speed = total_distance / frame_count
    print("Total Displacement:", total_distance)
    print("Average Speed:", average_speed, "pixels/frame")

cv2.destroyAllWindows()
```

```python
import cv2
import time

cap = cv2.VideoCapture("video.mp4")

ret, old_frame = cap.read()
old_gray = cv2.cvtColor(old_frame, cv2.COLOR_BGR2GRAY)

while True:

    ret, frame = cap.read()

    if not ret:
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # Lucas-Kanade
    start_lk = time.time()

    p0 = cv2.goodFeaturesToTrack(
        old_gray,
        maxCorners=100,
        qualityLevel=0.3,
        minDistance=7
    )

    if p0 is not None:
        p1, status, error = cv2.calcOpticalFlowPyrLK(
            old_gray,
            gray,
            p0,
            None
        )

    lk_time = time.time() - start_lk

    # Farneback
    start_fb = time.time()

    flow = cv2.calcOpticalFlowFarneback(
        old_gray,
        gray,
        None,
        0.5,
        3,
        15,
        3,
        5,
        1.2,
        0
    )

    fb_time = time.time() - start_fb

    print("Lucas-Kanade Time:", lk_time)
    print("Farneback Time:", fb_time)

    old_gray = gray.copy()

    if cv2.waitKey(30) & 0xFF == 27:
        break

cap.release()
cv2.destroyAllWindows()
```
